## 4. Backfill all four leagues

This cell initializes its own configuration, loads `.env`, and imports the collector. Run the dependency-install cell once if this Python environment does not yet have the project dependencies; the other collection/configuration cells can be skipped.

Set `BATCH_SEASON` inside this cell. `BATCH_END_WEEK = None` retrieves through each league's reported end week; set it to `5` to stop at week 5. Existing snapshots are skipped, so deleted player-data files will be retrieved again. Each league's results are saved in `all_backfill_statuses`; failures are listed in `backfill_errors`. Rate-limit responses stop the batch; other league failures allow the remaining leagues to proceed.


```python
# Standalone batch setup. Run the dependency-install cell once per environment.
from pathlib import Path
import sys

from dotenv import load_dotenv

BATCH_SEASON = 2025
BATCH_PROJECT_CANDIDATES = (Path.cwd().resolve(), Path.cwd().resolve() / "yahoo-fantasy-data")
BATCH_PROJECT_DIR = next(
    (path for path in BATCH_PROJECT_CANDIDATES if (path / "src/yahoo_fantasy_data").is_dir()),
    None,
)
if BATCH_PROJECT_DIR is None:
    raise RuntimeError("Open this notebook from the repository root or yahoo-fantasy-data directory.")

batch_env_file = BATCH_PROJECT_DIR / ".env"
if not batch_env_file.is_file():
    raise FileNotFoundError(f"Missing {batch_env_file}. Copy .env.example and fill it in.")
load_dotenv(batch_env_file, override=False)
batch_src_dir = str(BATCH_PROJECT_DIR / "src")
if batch_src_dir not in sys.path:
    sys.path.insert(0, batch_src_dir)

from yahoo_fantasy_data.config import load_settings
from yahoo_fantasy_data.yahoo import backfill_season
from yahoo_fantasy_data.errors import YahooRateLimitError

from dataclasses import replace

# Public-only batch: do not reuse OAuth tokens from .env or notebook memory.
batch_settings = replace(
    load_settings(data_dir=BATCH_PROJECT_DIR / "data"),
    client_id=None, client_secret=None, refresh_token=None,
)
print("Public API only; OAuth fallback is disabled.")

# Run this cell when you are ready to backfill all listed leagues.
LEAGUES_TO_BACKFILL = {
    "PHFFL_A": "707737",
    "CFFL_A": "134317",
    "CFFL_B": "918145",
    "Ferda": "889216",
}
BATCH_START_WEEK = 1
BATCH_END_WEEK = None  # Set to 5 to retrieve only weeks 1 through 5.
BATCH_OVERWRITE = False  # Keep existing snapshots; retrieve missing files.

all_backfill_statuses = {}
backfill_errors = {}

for nickname, league_id in LEAGUES_TO_BACKFILL.items():
    print(f"Backfilling {nickname} ({league_id}), season {BATCH_SEASON}...")
    try:
        result = backfill_season(
            season=BATCH_SEASON,
            league_id=league_id,
            start_week=BATCH_START_WEEK,
            end_week=BATCH_END_WEEK,
            overwrite=BATCH_OVERWRITE,
            settings=batch_settings,
            league_nickname=nickname,
        )
        all_backfill_statuses[nickname] = result
        status_counts = {}
        for week_statuses in result.values():
            for status in week_statuses.values():
                status_counts[status] = status_counts.get(status, 0) + 1
        print(f"  Processed {len(result)} weeks: {status_counts}")
    except YahooRateLimitError as exc:
        backfill_errors[nickname] = str(exc)
        print("Yahoo rate-limited the request. Stopping the batch; retry later.")
        break
    except Exception as exc:
        backfill_errors[nickname] = str(exc)
        print(f"  Failed: {type(exc).__name__}. See backfill_errors for details.")

print(f"Finished: {len(all_backfill_statuses)} leagues processed, {len(backfill_errors)} failed.")
all_backfill_statuses
```
